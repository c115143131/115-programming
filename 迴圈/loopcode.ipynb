# Loop

Loop 指令可以讓程式碼依我的條件執行，直到條件滿足為止

```python
#底下程式碼會印出1~9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i+1)
```

```python
sum = 0
for i in range(10):
    sum = sum + (i+1)
print(sum)
```

# 印出圖形

```python
for i in range(6):
    print("&"*(i+1))
```

印出  
1 * 1 = 1  
1 * 2 = 2  
1 * 3 = 3  
1 * 4 = 4  
1 * 5 = 5  

```python
print(f'1*1={1*1}')
print(f'1*2={1*2}')
print(f'1*3={1*3}')
print(f'1*4={1*4}')
print(f'1*5={1*5}')
```

```python
for i in range(10):
    print(f'1*{(i+1)}={(1*i+1)}')
```

```python
for i in range(10):
    print(f'{(i+1)}*2={(i+1)*2}')
```
