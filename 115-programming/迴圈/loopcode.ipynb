# Loop

Loop 指令可以讓程式碼依我的條件執行，直到條件滿足為止

```python
#底下程式碼會印出1~9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i+1)
```

```python
sum = 0
for i in range(10):
    sum = sum + (i+1)
print(sum)
```
